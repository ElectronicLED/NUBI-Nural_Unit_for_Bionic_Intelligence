### Speech to Text using OpenAI Whisper Model 

**Requirements**
-   Download Python 3.13.9 with whisper model 
        [✔️]
-   OpenAI Python Lib       [✔️]
-   Pathlib Python Lib      [✔️]

```python
# How to use whisper for Speech to Text
import whisper
from pathlib import Path

# Transcribing an audio file
model = whisper.load_model("base")
audio_path = r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\1-Audio_Recording\recorded_input.wav"
text = model.transcribe(audio_path)
text = text['text']

# Saving the transcribed text to a file
out_path = Path(r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\2-STT_TTS\STT_Output.txt")
out_path.write_text(text, encoding="utf-8")

```

### Text to Speech using (Google Text-to-speech)

```python
import sys
print(sys.executable)  # Print the path of the current Python interpreter

from gtts import gTTS  # Google Text-to-Speech
from pathlib import Path

# Read the text from your file
text_path = Path(r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\5-API Behaviour\ai_reply.txt")
text = text_path.read_text(encoding="utf-8")

# Convert to speech
tts = gTTS(text)
tts.save(r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\2-STT_TTS\Robot_Response.wav")
```
