```python
import os
import cohere

# === CONFIG ===
INPUT_DIR = r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\2-STT_TTS"      # directory where speech.txt lives
OUTPUT_DIR = r"C:\Users\MichelleGh\Desktop\Graduation Project\Verbal Communication\5-API Behaviour"       # directory where ai_reply.txt will be saved

INPUT_FILE_NAME = "STT_Output.txt"
OUTPUT_FILE_NAME = "ai_reply.txt"

COHERE_API_KEY = "pxbp77BALEELCHPMtk6IGPv3iejrSUpwYFGWm5ML"

# === INIT COHERE ===
co = cohere.Client(api_key=COHERE_API_KEY)


def process_text_with_cohere(text):
    response = co.chat(
        model="command-a-03-2025",
        message=text
    )
    return response.text


def main():
    input_path = os.path.join(INPUT_DIR, INPUT_FILE_NAME)
    output_path = os.path.join(OUTPUT_DIR, OUTPUT_FILE_NAME)

    # Ensure input file exists
    if not os.path.exists(input_path):
        print(f"Input file not found: {input_path}")
        return

    # Read input text
    with open(input_path, "r", encoding="utf-8") as f:
        input_text = f.read().strip()

    if not input_text:
        print("Input file is empty. Exiting.")
        return

    print("Sending to Cohere...")
    ai_response = process_text_with_cohere(input_text)

    # Ensure output directory exists
    os.makedirs(OUTPUT_DIR, exist_ok=True)

    print("Saving response...")
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(ai_response)

    print("Done. Response saved to:", output_path)


if __name__ == "__main__":
    main()
```
